Creamos un pequeño programa que descuente el inventario de una hamburguesa

```python
# Inicializamos la variable 'inventario' con un valor inicial de 100 hamburguesas
inventario = 100

# Mostramos al usuario el mensaje inicial con la cantidad actual de hamburguesas
print(f"El inventario contiene {inventario} hamburguesas")

# Iniciamos un bucle que se ejecutará repetidamente MIENTRAS el inventario sea mayor que 0
while inventario > 0:

    # Verificamos si quedan 10 o menos hamburguesas para lanzar una alerta preventiva
    if inventario <= 10:
        print("El inventario esta casi vacio: CUIDADO")

    # Solicitamos al usuario la cantidad que quiere el cliente, la convertimos a un número entero (int) y la guardamos
    num_hamburguesas = int(input("Cuantas hamburguesas quiere el cliente: "))

    # Si el cliente pide más hamburguesas de las que tenemos disponibles en el inventario...
    if num_hamburguesas > inventario:
        # ...le avisamos que no es posible completar el pedido y le decimos cuántas quedan
        print("No hay suficientes hamburguesas. Solo hay", inventario, "hamburguesas disponibles")

    # Si hay suficientes hamburguesas para cubrir el pedido...
    else:
        # Restamos las hamburguesas vendidas del total del inventario (es lo mismo que: inventario = inventario - num_hamburguesas)
        inventario -= num_hamburguesas
        # Confirmamos la venta y mostramos cuántas hamburguesas quedan en tiempo real
        print(f"El cliente ha pedido {num_hamburguesas} hamburguesas. El inventario ahora tiene {inventario} hamburguesas")
```
