```python
%%html
<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Roci's Cyber Guessing Game 🔮</title>
    <style>
        /* 🎨 EL LOOK & FEEL: Estilo Cyberpunk / Arcade */
        body {
            font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
            background: linear-gradient(135deg, #0f2027, #203a43, #2c5364);
            color: #ffffff;
            display: flex;
            justify-content: center;
            align-items: center;
            height: 100vh;
            margin: 0;
        }

        .game-card {
            background: rgba(255, 255, 255, 0.1);
            backdrop-filter: blur(10px);
            padding: 30px;
            border-radius: 20px;
            box-shadow: 0 8px 32px 0 rgba(0, 0, 0, 0.37);
            border: 1px solid rgba(255, 255, 255, 0.18);
            text-align: center;
            max-width: 400px;
            width: 100%;
        }

        h1 {
            color: #00ffcc;
            text-shadow: 0 0 10px #00ffcc;
            margin-bottom: 5px;
        }

        .creator {
            font-size: 0.9rem;
            color: #ff007f;
            margin-bottom: 25px;
            font-weight: bold;
        }

        .hearts {
            font-size: 1.5rem;
            margin-bottom: 20px;
            letter-spacing: 5px;
        }

        input[type="number"] {
            width: 80px;
            padding: 10px;
            font-size: 1.5rem;
            border-radius: 10px;
            border: 2px solid #00ffcc;
            background: transparent;
            color: white;
            text-align: center;
            outline: none;
        }

        button {
            background: #ff007f;
            color: white;
            border: none;
            padding: 12px 25px;
            font-size: 1.1rem;
            border-radius: 10px;
            cursor: pointer;
            font-weight: bold;
            margin-left: 10px;
            transition: all 0.3s ease;
            box-shadow: 0 0 15px rgba(255, 0, 127, 0.4);
        }

        button:hover {
            background: #00ffcc;
            color: #0f2027;
            box-shadow: 0 0 15px rgba(0, 255, 204, 0.8);
        }

        .message {
            margin-top: 25px;
            font-size: 1.2rem;
            font-weight: bold;
            min-height: 50px;
        }

        .restart-btn {
            display: none;
            background: #00ffcc;
            color: #0f2027;
            margin: 20px auto 0 auto;
        }
    </style>
</head>
<body>

    <div class="game-card">
        <h1>CYBER GUESS</h1>
        <div class="creator">By Roci (The Developer) 🚀</div>

        <p>El sistema ha pensado un número del <b>0 al 9</b>.<br>¿Puedes hackear su mente?</p>

        <div class="hearts" id="heartsContainer">💖 💖 💖</div>

        <div id="inputArea">
            <input type="number" id="userGuess" min="0" max="9" placeholder="?">
            <button onclick="comprobarIntento()">¡PROBAR!</button>
        </div>

        <div class="message" id="feedbackMessage">¡Introduce tu predicción y pulsa el botón!</div>

        <button class="restart-btn" id="restartBtn" onclick="reiniciarJuego()">Volver a Jugar 🔄</button>
    </div>

    <script>
        // Variables de estado globales
        let numeroMaquina;
        let vidasRestantes;
        let juegoTerminado;

        // Función que inicializa o resetea la partida
        function reiniciarJuego() {
            numeroMaquina = Math.floor(Math.random() * 10); // Genera de 0 a 9
            vidasRestantes = 3;
            juegoTerminado = false;

            // Actualizar la interfaz visual
            document.getElementById("heartsContainer").innerHTML = "💖 💖 💖";
            document.getElementById("feedbackMessage").innerHTML = "¡Nueva partida iniciada! Intenta adivinar.";
            document.getElementById("feedbackMessage").style.color = "#ffffff";
            document.getElementById("userGuess").value = "";
            document.getElementById("inputArea").style.display = "block";
            document.getElementById("restartBtn").style.display = "none";

            console.log("🤫 Trampa para el profe, el número es: " + numeroMaquina);
        }

        // Función principal que se ejecuta al pulsar el botón
        function comprobarIntento() {
            if (juegoTerminado) return;

            const inputElement = document.getElementById("userGuess");
            const userNumber = parseInt(inputElement.value);
            const messageElement = document.getElementById("feedbackMessage");

            // ⚠️ Validación: Control de errores
            if (isNaN(userNumber) || userNumber < 0 || userNumber > 9) {
                messageElement.innerHTML = "⚠️ ¡Ey, Roci! Introduce un número válido entre 0 y 9.";
                messageElement.style.color = "#ffcc00";
                return;
            }

            // Si el dato es válido, procesamos la lógica
            if (userNumber === numeroMaquina) {
                // 🎉 ¡VICTORIA!
                messageElement.innerHTML = "🎉 ¡BRUTAL ROCI! ¡Has ganado! El sistema ha sido hackeado. 😎";
                messageElement.style.color = "#00ffcc";
                finalizarPartida();
            } else {
                // ❌ ¡FALLO! Restamos una vida
                vidasRestantes--;
                actualizarVidasVisuales();

                if (vidasRestantes > 0) {
                    // Pistas dinámicas
                    if (userNumber < numeroMaquina) {
                        messageElement.innerHTML = `❌ ¡Fallo! El número secreto es <b>MÁS ALTO</b> que ${userNumber}.`;
                    } else {
                        messageElement.innerHTML = `❌ ¡Fallo! El número secreto es <b>MÁS BAJO</b> que ${userNumber}.`;
                    }
                    messageElement.style.color = "#ffffff";
                } else {
                    // 💥 GAME OVER
                    messageElement.innerHTML = `💥 ¡GAME OVER! Has agotado tus vidas. El número era el <b>${numeroMaquina}</b>. 🤖`;
                    messageElement.style.color = "#ff007f";
                    finalizarPartida();
                }
            }

            inputElement.value = ""; // Limpiar el input para el siguiente intento
        }

        // Dibuja los corazones según las vidas que queden
        function actualizarVidasVisuales() {
            let corazones = "";
            for (let i = 0; i < 3; i++) {
                if (i < vidasRestantes) {
                    corazones += "💖 ";
                } else {
                    corazones += "🖤 "; // Corazón roto/vacío
                }
            }
            document.getElementById("heartsContainer").innerHTML = corazones;
        }

        // Esconde los controles de juego y muestra el botón de reinicio
        function finalizarPartida() {
            juegoTerminado = true;
            document.getElementById("inputArea").style.display = "none";
            document.getElementById("restartBtn").style.display = "block";
        }

        // Arrancar el juego por primera vez al cargar la web
        reiniciarJuego();
    </script>
</body>
</html>
```
